```python
df = pd.read_csv("supermarket_sales_dirty.csv")
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

df['Age'].fillna(df['Age'].mean(), inplace=True)
```

```python
df['Purchase_Amount'].fillna(
    df['Purchase_Amount'].median(),
    inplace=True
)
```

```python
df['City'].fillna("Unknown", inplace=True)
```

```python
df['Purchase_Date'].fillna(
    "2024-01-01",
    inplace=True
)
```

```python
df.duplicated()
```

```python
df.duplicated().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['Name'] = df['Name'].str.strip()

df['City'] = df['City'].str.strip()
```

```python
df['City'] = df['City'].str.title()
```

```python
df[df['Age'] > 100]
```

```python
df.loc[df['Age'] > 100, 'Age'] = df['Age'].median()
```

```python
df[df['Purchase_Amount'] < 0]
```

```python
df.loc[
    df['Purchase_Amount'] < 0,
    'Purchase_Amount'
] = abs(
    df['Purchase_Amount']
)
```

```python
df['Purchase_Date'] = pd.to_datetime(
    df['Purchase_Date']
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(df.isnull(), cbar=False, cmap="Reds")

plt.title("Missing Values Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))

df['City'].value_counts().plot(
    kind='bar',
    color='skyblue'
)

plt.title("Customers by City")
plt.xlabel("City")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,6))

df['Gender'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Gender Distribution")
plt.ylabel('')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['Purchase_Amount'],
    bins=10,
    kde=True,
    color='green'
)

plt.title("Purchase Amount Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='Product',
    data=df,
    palette='viridis'
)

plt.title("Product Sales Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='Product',
    y='Purchase_Amount',
    data=df
)

plt.title("Purchase Amount by Product")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['Age'],
    bins=8,
    color='orange'
)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.show()
```

```python
df['Purchase_Date'] = pd.to_datetime(df['Purchase_Date'])

monthly_sales = df.groupby(
    df['Purchase_Date'].dt.month
)['Purchase_Amount'].sum()

plt.figure(figsize=(8,5))

monthly_sales.plot(
    marker='o',
    linewidth=3
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(6,4))

numeric_df = df[['Age','Purchase_Amount']]

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
fig, axes = plt.subplots(2,2, figsize=(14,10))

# City Count
df['City'].value_counts().plot(
    kind='bar',
    ax=axes[0,0],
    color='skyblue'
)
axes[0,0].set_title("Customers by City")

# Product Count
sns.countplot(
    x='Product',
    data=df,
    ax=axes[0,1]
)
axes[0,1].set_title("Products Sold")

# Age Distribution
sns.histplot(
    df['Age'],
    ax=axes[1,0],
    color='orange'
)
axes[1,0].set_title("Age Distribution")

# Purchase Amount
sns.histplot(
    df['Purchase_Amount'],
    ax=axes[1,1],
    color='green'
)
axes[1,1].set_title("Purchase Distribution")

plt.tight_layout()
plt.show()
```
